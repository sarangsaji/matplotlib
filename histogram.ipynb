HISTOGRAM


A histogram is a graphical representation of distribution of the numerical data.it displays data using contigues bars for to show the frequency of data points within specified ranges of intervals.histogram is particularly useful for understanding the underlaying frequency distribution(skewed distribution) of a data set

COMPONENTS OF HISTOGRAM

bins(or intervals) data range is divided into consicutive non_overlapping intervals the bins are typically of equal width but they can vary if needed bars:each bar represents the count or frequency of data points within a bin.the height of the bar indicates.how many data points fail into each interval

x-axis represents the range of values in the data,divided into bins

y-axis represents the frequency of relative frequency of data points within each bin

```python
import matplotlib.pyplot as plt
import numpy as np
a=np.array([22,87,12,18,71,55,44,
            11,21,51,5,78,11,27])
plt.hist(a,color="r",bins=20)
plt.title("HISTOGRAM")
```
