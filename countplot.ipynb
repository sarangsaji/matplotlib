Countplot

A countplot is a cstegorical plot used to show the frequency (count) of observations in each category of a dataset.
it is available in seaborn and is one of the most commonly used plots in explorarity data analysis (EDA).

What is countplot shows

```python
# A countplot answers questions like:
# how many males and females  are in thee dataset:
# how many students passed vs failed:
# how many orders came each city:
# in simple words:
# it counts how many times each category appears:
```

LOADING AND DISPLAYING A SAMPLE DATA SET

```python
import pandas
import matplotlib
import seaborn as sns
df=sns.load_dataset("iris")
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
#loading dataset
data=sns.load_dataset("iris")
#creating a count plot
sns.countplot(x="species",data=data)
#showing the plot
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
#loading dataset
data=sns.load_dataset("iris")
#creating a count plot
sns.countplot(y="species",data=data)
#showing the plot
plt.show()
```

CHANGING COLORS

column palettes:set1,set2,coolwarm,pastel,dark

```python
sns.countplot(x="species" , data=data,palette="coolwarm")
plt.show()
```

ORDERING CATEGORIES

```python
sns.countplot(x="species",data=data,order=["setosa","versicolor","virginica"])
plt.show()
```
