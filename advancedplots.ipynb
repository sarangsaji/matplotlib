VIOLIN PLOT

Violin plot is a similiar to a box plot but provides a more detailed view of the distribution by including a kernal density estimation which smooths the distribution curve

```python
import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset("iris")
data
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset('tips')
sns.stripplot(x='day', y='total_bill', data=data,
              hue='sex', palette='Set1')
plt.show()
```

STRIP PLOT

A strip plot is a simple scatter plot for categorical data it plots the data points individually the summarizing them into averages

```python
#loading data set
data=sns.load_dataset("iris")
#creating a stripplot
sns.stripplot(x="species",y="sepal_width",data=data)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')  

sns.violinplot(x="day", y="total_bill",
               data=data, hue='sex', split=True, palette='Set1')

plt.show()
```
