KDE PLOT(kdeplot)

A kernal density estimate(KDE) plot represents the distribution of a continious variable it smooths out the distribution using a continious probability curve.

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset("tips")
```

```python
sns.kdeplot(tips["total_bill"])
```

HEATMAP(heatmap)

A heatmap is a matrix plot that visualizes the values in a matrix using color coding its particularly useful for visualizing coorelations in a dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = sns.load_dataset("tips")
#coorelation matrix heatmap
tc = data.corr(numeric_only=True)
sns.heatmap(tc,annot=True,cmap='plasma',linecolor='black',linewidths=1)
plt.show()
```

JOINTPLOT(jointplot)

the jointplot() combines scatter plots and histogram for bivariate analysis (two variables). it allows you to visualize the relationships between two variables along with the univeritate distributions  

```python
sns.jointplot(x='total_bill',y='tip',data=tips)
plt.grid()
```
