LINE PLOT-MONTHLY SALES

```python
import matplotlib.pyplot as plt
import numpy as np
x=["jan","feb","mar","apr","may","jun","jul","aug","sep","oct","nov","dec"]
y=[20000,22000,25000,24000,28000,30000,32000,31000,35000,37000,40000,45000]

plt.plot(x,y)
plt.title("MONTHLY SALES DATA")
plt.xlabel("MONTHS")
plt.ylabel("SALES")
plt.grid("True")
plt.show()
```

```python
y=[20000,22000,25000,24000,28000,30000,32000,31000,35000,37000,40000,45000]
total=sum(y)
print("total yearly sales:",total)
```

```python
plt.plot(x, y, marker='o', linestyle='--', linewidth=2)
plt.grid("True")
```

BAR CHART--STUDENTS IN DEPARTMENTS

```python
import matplotlib.pyplot as plt
departments=["BCA","BBA","Bcom","BA","BSc"]
students=[120,90,140,80,110]
colors=["yellow","hotpink","green","black","blue"]
plt.bar(departments,students,color=colors,width=0.5)
plt.xlabel("DEPARTMENTS")
plt.ylabel("STUDENTS")
plt.title("STUDENTS STRENGTH")
plt.show()
```

```python
import matplotlib.pyplot as plt
departments=["BCA","BBA","Bcom","BA","BSc"]
students=[120,90,140,80,110]
colors=["yellow","hotpink","green","black","blue"]
plt.barh(departments,students,color=colors)
plt.xlabel("DEPARTMENTS")
plt.ylabel("STUDENTS")
plt.title("STUDENTS STRENGTH")
plt.show()
```

BOX PLOT

```python
import seaborn as sns 
import matplotlib.pyplot as plt
salary=[25000,26000,27000,28000,29000,30000,31000,50000,27000,26000]
sns.boxplot(data=salary)
plt.title("SALARY BOXPLOT")
plt.show()
```

COUNTPLOT

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data={"gender":["male","female","female","male","male","female","male","female","male","female"]}
df=pd.DataFrame(data)
sns.countplot(x="gender",data=df,palette="coolwarm")
plt.show()
```
