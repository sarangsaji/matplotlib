PIE CHART

SIMPLE PIE CHART

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([35,25,25,25])
plt.pie(x)
plt.show()
```

PIE CHART WITH LABELS

```python
y=np.array([35,25,25,15])
my_labels=["apple","banana","mango","strawberry"]
plt.pie(y,labels=my_labels)
plt.show()
```

PIE CHART WITH START ANGLE

```python
y=np.array([35,25,25,15])
my_labels=["apple","banana","mango","strawberry"]
plt.pie(y,labels=my_labels,startangle=90)
plt.show()
```

EXPLODED PIE CHART

```python
y=np.array([35,25,25,15])
my_labels=["apple","banana","mango","strawberry"]
my_explode=[0,0.2,0,0]
plt.figure(figsize=(4,4))
plt.pie(y,labels=my_labels,explode=my_explode)
plt.show()
```

```python
y=np.array([35,25,25,15])
my_labels=["apple","banana","mango","strawberry"]
my_explode=[0,0.2,0,0]
plt.pie(y,labels=my_labels,explode=my_explode,shadow=True)
plt.show()
```

PIE CHART WITH CUSTOM COLORS

```python
import matplotlib.pyplot as plt
import numpy as np
y=np.array([35,25,25,15])
my_labels=["apple","banana","mango","strawberry"]
my_colors=["#ca4e4e","hotpink","green","orange"]
plt.pie(y,labels=my_labels,colors=my_colors)
plt.show()
```

PIE CHART WITH LEGEND

```python
y=np.array([35,25,25,15])
my_labels=["apple","banana","mango","strawberry"]
plt.pie(y,labels=my_labels)
plt.legend()
plt.show()
```
