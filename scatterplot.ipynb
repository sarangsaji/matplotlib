What is a scatter plot

A scatter plot is a type of graph that shows the relationship between two numerical(continious) variables.

each dot (point) represents one observations

the x-axis one numerical variable

the y axis another numerical variable

it helps us understand patterns,trends,relationships,and outlieers in data

why scatter plots are important (especially in data science)



scatter plots helps you

identify relationships between variables

detect correlation (positive,negative,none)

find outliers(unusual data points)

decide whether machine learning models like regression can be applied

```python
import matplotlib.pyplot as plt
x=[5,7,8,7,2,17,2,9,4,11,12,9,6]
y=[99,86,87,88,111,86,103,87,84,78,77,85,86]
plt.scatter(x,y)
plt.show()
```

colors

you can set your own color for each scatter plot with the color or the c argument

```python
import matplotlib.pyplot as plt

x=[5,8,7,8,7,2,17,2,9,4,11,12,9,6]
y=[99,86,87,88,111,86,103,87,94,78,77,85,86,87]
plt.scatter(x,y,color='hotpink')

x=[2,2,8,1,15,8,12,9,7,3,11,4,7,14]
y=[105,84,105,90,99,90,95,94,100,79,112,91,80,85]
plt.scatter(x,y,color='#4287f5')

plt.show()
```

color each dot

```python
import matplotlib.pyplot as plt

x=[5,8,7,8,7,2,17,2,9,4,11,12,9,6]
y=[99,86,87,88,111,86,103,87,94,78,77,85,86,87]
colors=["red","green","blue","yellow","pink","black","orange","purple","beige","brown","gray","cyan","magenta","violet"]
plt.scatter(x,y,c=colors)
plt.show()
```

colormap

the matplotlib module has a number of valiable colormaps

a color map like a list of colors where each color has a value than ranges from 0 to 100

how to use the colormap

you can specify thne colormap with the keyword arguement with value of thne colormap,in this case"virdis" one of the  built in colormaps avaiable in matplotlib 

```python
import matplotlib.pyplot as plt
import numpy as np
x=[5,7,8,7,2,17,2,9,4,11,12,9,6]
y=[90,86,87,111,103,94,78,77,85,90,100,85,70]
colors=[0,70,56,67,8,22,66,88,50,23,14,56,76]

plt.scatter(x,y, c=colors, cmap ="viridis")
plt.show()
```

size

you can change the size of thee dots with the s argument.

just like colors,make sure the array for sizes has the same length as the arrays for the x-axis and y-axis

```python
import matplotlib.pyplot as plt 
x=[5,7,8,7,2,17,2,9,4,11,12,9,6]
y=[99,86,87,88,111,86,103,87,94,78,77,85,86]

plt.scatter(x,y,s=50)
plt.show()
```

STUDY HOURS VS MARKS

```python
import matplotlib.pyplot as plt
study_hours=[1,2,3,4,5,6,7,8]
marks=[35,45,55,65,75,85,95,100]

plt.scatter(study_hours,marks)
plt.xlabel("study hours")
plt.ylabel("marks")
plt.title("study hours vs marks")
plt.show()
```

negative correlation

```python
days_before_exam=[30,25,20,15,10,5]
study_hours=[1,2,3,4,6,8]
plt.scatter(days_before_exam,study_hours)
plt.xlabel("days before exam")
plt.ylabel("study hours per day")
plt.ylabel("study hours per day")
plt.title("negative correlation distance vs study hours")
plt.show()
```

SCATTER PLOT WITH ALPHA (TRANSPERENCY)

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y=np.array([90,86,87,88,111,86,101,87,94,78,77,85,80])
sizes=np.array([20,50,100,200,500,1000,60,90,10,300,600,800,75])
plt.scatter(x,y,s=sizes,alpha=1)
plt.show()
```
