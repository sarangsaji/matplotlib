#  what is bar chart?

a bar chart for bar graphis used to compare values across different categories.

each bar represents a category

the height (or lenght) of the bar show the values


categories are usually discrete(not continuous)

# Bar chart = category comparison

```python
#component                           meaning

# x-axis                             categories(names/labels)
# y-axis                             numerical values
# bars                               represent data values
# height of bar                      magnitude of the value
# Title                              what the chart represent?
# labels                             meaning of axes
```

```python
import matplotlib.pyplot as plt
x=["A","B","C","D"]
y=[3,8,1,10]
plt.bar(x,y)
plt.show()
```

The bar() function takes arguments that describes the layout of then bars 

the categories and their values represented by the first and second argument as list

HORIZONTAL BARS

If you want the bars to be displayed horizontally instead of veritically use the barh() functions

```python
import matplotlib.pyplot as plt
x=["A","B","C","D"]
y=[3,8,1,10]
plt.barh(x,y)
plt.show()
```

BAR COLOR

The bar() and barh() take the keyword color to ser the color of the bars:

```python
x=["A","B","C","D"]
y=[3,8,1,10]
plt.bar(x,y,color="black")
plt.show()
```

BAR WIDTH

The bar() takes the keyword argument width to set the width of the bars

The default width value is 0.8

```python
import matplotlib.pyplot as plt 
x=["A","B","C","D"]
y=[3,8,1,10]
plt.bar(x,y,width=0.2,color="black")

plt.show()
```

BAR HEIGHT

For horizontal bars,use height instead of widcth.

the bath() takes the keyword argument height to set the height of the bars

```python
x=["A","B","C","D"]
y=[3,8,1,10]
plt.barh(x,y,height=0.3,color="black")
plt.show()
```

REAL-TIME EXAMPLES

1:STUDENTS IN EACH DEPARTMENT

A college wants to compare number of students in each department 

Department students CS 120 IT 90 ECE 70 EEE 60

```python
import matplotlib.pyplot as plt 
departments=["CS","IT","ECE","EEE"]
students=[120,90,70,60]
clr=["red","pink","green","yellow"]
plt.bar(departments,students,color=clr,width=0.5)
plt.xlabel("DEPARTMENT")
plt.ylabel("NUMBER OF STUDENTS")
plt.title("DEPARTMENT WISE STUDENTS STRENGTH")
plt.show()
```

2.EXAMPLE COUNT BY ROLE

```python
roles=["MANAGER","DEVOLOPER","TESTER","HR"]
count=[5,20,8,4]
plt.barh(roles,count,height=0.4,color="purple")
plt.xlabel("NUMBER OF EMPLOYEES")
plt.ylabel("JOB ROLE")
plt.title("EMPLOYEE DISTRIBUTION")
plt.show()
```

MULTIPLE BAR CHART(GROUPED BAR CHART)

```python
Compare boys vs girls in each class.
class       boys      girls
10          30        25 
11          28        27
12          26        29
```

```python
import numpy as np
classes=["class_10","class_11","class_12"]
boys=[30,28,26]
girls=[25,27,29]

x=np.arange(len(classes))
plt.bar(x+0.2,boys,width=0.4,label="boys")
plt.bar(x-0.2,girls,width=0.4,label="girls")
plt.xticks(x,classes)
plt.xlabel("class")
plt.ylabel("number of students")
plt.title("boys vs girls strength")
plt.legend()
plt.show()
```
