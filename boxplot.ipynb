# Boxplot (box-and whisker plot)

## what is boxplot?

a boxplot is a statistical graph that summarizes the distribution of a numerical dataset using five key values it shows:

central value()

## why do we use boxplot?

## problems with raw data/mean alone:

mean hides variation mean is affected by outliers hard to compare groups boxplot solves this by:

using median(robust)showing spread(IQR)highlighting outliers easy comparison between groups.a boxplot is built from five numbers:

```python
# component                       meaning 
# minimum                         smallest value(excluding outliers)
# Q1                              25% of data below
# median(Q2)                      50% of data
# Q3                              75% of data below
# maximum                         largest value(excluding outliers)
```

quartiles-how data is divided when data is sorted

Q1-value at 25% median-value at 50% q3-value at 75%

so the data is divided into four equal parts

structure of boxplot

min-Q1-median-Q3-max

```python
import seaborn as sns
import matplotlib.pyplot as plt
data1=[2,45,50,55,60,65,70,80,120]
sns.boxplot(data=data1)
plt.title("SIMPLE BOXPLOT")
plt.show()
```

HORIZONTAL BOXPLOT

```python
import seaborn as sns
import matplotlib.pyplot as plt
data1=[2,45,50,55,60,65,70,80,120]
sns.boxplot(data=data1,orient='h')
plt.title("HORIZONTAL BOX PLOT")
plt.show()
```

median line

verticle line inside the box,divides data into two equal halves,preffered over mean because its not affected by outliers

interquatile range(IQR)


IQR=Q3-Q1,measures spread of middle 50% robort measures of variability

whiskers

whiskers extend from Q1-Lower limit Q3-Upper limit

lower limit=Q1-1.5*IQR

Upper limit=Q3+1.5*IQR

OUTLIERS

Values outside whiskers

shows as dots

outliers are unusual values,not neccesarily errors

COMPARING TWO GROUPS

```python
group_A=[60,65,70,75,80]
group_B=[55,60,62,68]
sns.boxplot(data=[group_A,group_B])
plt.xticks([0,1],["group_A","group_B"])
plt.title("group comparison using boxplot")
plt.show()
```
